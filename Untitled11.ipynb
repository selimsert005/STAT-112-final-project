```python
import pandas as pd
import numpy as np

# Configuration
FILE_PATH = 'sleep_health_lifestyle_synthetic_1200_dirty_v2.csv'
OUTPUT_FILE = 'final_cleaned_sleep_data.csv'

# 1. Load Data
try:
    df = pd.read_csv(FILE_PATH)
    print("Data loaded successfully.")
    print(f"Initial row count: {len(df)}")
except FileNotFoundError:
    print("ERROR: File not found! Please check the filename and ensure it is uploaded.")
    df = pd.DataFrame()

if not df.empty:
    # 2. Clean Column Names
    df.columns = df.columns.str.replace(r'[!#?~]', '', regex=True).str.strip().str.lower()

    # 3. Remove Duplicate Rows
    initial_count = len(df)
    df = df.drop_duplicates()
    print(f"Duplicate rows removed: {initial_count - len(df)}")

    # 4. Clean String (Categorical) Data
    categorical_cols = ['gender', 'occupation', 'bmi_category', 'smoking_status', 'sleep_disorder']
    present_categorical = [col for col in categorical_cols if col in df.columns]

    for col in present_categorical:
        df[col] = df[col].astype(str)
        df[col] = df[col].str.strip().str.replace(r'[~!#?*]', '', regex=True)
        df[col] = df[col].str.lower()
        df[col] = df[col].replace({'nan': np.nan, 'na': np.nan, 'none': np.nan, 'null': np.nan})

    # 5. Correct Typos
    corrections = {
        'gender': {'femle': 'female', 'femal': 'female', 'mal': 'male'},
        'occupation': {
            'software  engineer': 'software engineer',
            'healthcare  worker': 'healthcare worker', 'healthcare workr': 'healthcare worker',
            'managr': 'manager', 'freelancr': 'freelancer'
        },
        'bmi_category': {
            'nrmal': 'normal', 'undrweight': 'underweight', 'underwight': 'underweight',
            'nderweight': 'underweight', 'obse': 'obese'
        },
        'work_shift': {'rotating/shft': 'rotating/shift', 'nght': 'night', 'dy': 'day'},
        'smoking_status': {'formr': 'former', 'nver': 'never', 'nevr': 'never', 'non': 'none', 'nne': 'none'},
        'sleep_disorder': {'insomni': 'insomnia', 'sleep apne': 'sleep apnea', 'slep apnea': 'sleep apnea', 'sleep  apnea': 'sleep apnea'}
    }

    for col, mapping in corrections.items():
        if col in df.columns:
            df[col] = df[col].replace(mapping)

    # 6. Clean Numerical Data
    numeric_cols = [
        'age', 'physical_activity_min_week', 'stress_score',
        'sleep_duration_hours', 'sleep_quality_score',
        'resting_heart_rate_bpm', 'daily_steps'
    ]
    present_numeric = [col for col in numeric_cols if col in df.columns]

    for col in present_numeric:
        if df[col].dtype == 'O':
            df[col] = df[col].astype(str).str.replace(r'[~!#?*]', '', regex=True)
        df[col] = pd.to_numeric(df[col], errors='coerce')

    # 7. Outlier Handling (Age & Others)

    if 'age' in df.columns:
        # Calculating mean
        mean_age = int(df['age'].mean())

        df['age'] = df['age'].fillna(mean_age)

        df.loc[(df['age'] < 18) | (df['age'] > 90), 'age'] = mean_age

        df['age'] = df['age'].astype(int)

    # Statistical Logic for other numeric columns
    other_numeric = [col for col in present_numeric if col != 'age']
    for col in other_numeric:
        if df[col].notnull().sum() > 0:
            mean_val = df[col].mean()
            std_val = df[col].std()
            upper_limit = mean_val + (3 * std_val)
            lower_limit = mean_val - (3 * std_val)

            outliers = (df[col] > upper_limit) | (df[col] < lower_limit)
            if outliers.sum() > 0:
                print(f"Correcting {outliers.sum()} outliers in column: {col}")
                df.loc[outliers, col] = mean_val

    # 8. Handle Missing Values (Rest of the data)
    missing_perc = df.isnull().mean()
    cols_to_drop = missing_perc[missing_perc > 0.60].index
    if not cols_to_drop.empty:
        df = df.drop(columns=cols_to_drop)

    for col in df.columns:
        if df[col].isnull().sum() > 0:
            if df[col].dtype in ['float64', 'int64']:
                df[col] = df[col].fillna(df[col].mean())
            else:
                if not df[col].mode().empty:
                    df[col] = df[col].fillna(df[col].mode()[0])

    # 9. Final formatting for integers
    int_candidates = ['daily_steps', 'physical_activity_min_week', 'resting_heart_rate_bpm']
    for col in int_candidates:
        if col in df.columns:
            df[col] = df[col].fillna(0).astype(int)

    # 10. Save Result
    print("\nCleaned Data Summary:")
    print(df.info())
    print("\nFirst 5 Rows:")
    print(df.head())

    df.to_csv(OUTPUT_FILE, index=False)
    print(f"\nSUCCESS: '{OUTPUT_FILE}' has been created and saved.")
```
